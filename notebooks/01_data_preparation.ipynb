Sanitised and reorganised from the original data-preparation notebook. Use only an authorised local workbook. No study records or outputs are distributed.

```python
from pathlib import Path
import pandas as pd

# Works from the repository root or notebooks/.
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data"
OUTPUT = ROOT / "outputs"
from sklearn.preprocessing import LabelEncoder
```

## Load and inspect the workbook locally

```python
INPUT = DATA / "renal_raw.xlsx"
df = pd.read_excel(INPUT)
display(df.dtypes)
# Avoid displaying raw records that may contain identifying information.
```

## Remove direct identifiers and encode categorical variables

The original workflow used independent label encoding for species, breed, sex and reproductive status. Codes are nominal labels, not ordered measurements. Review additional identifying columns before saving.

```python
df = df.drop(columns=["Name", "Index"], errors="ignore")
categorical_columns = ["Species", "Breed", "Sex", "Reproductive status"]
missing = set(categorical_columns) - set(df.columns)
if missing:
    raise ValueError(f"Missing required columns: {sorted(missing)}")

encoders = {}
for column in categorical_columns:
    encoder = LabelEncoder()
    df[column] = encoder.fit_transform(df[column])
    encoders[column] = encoder
```

## Local descriptive inspection and export

Keep all generated outputs private. The encoded workbook is an intermediate, not an anonymisation guarantee.

```python
display(df.describe())
display(df.dtypes)
DATA.mkdir(parents=True, exist_ok=True)
df.to_excel(DATA / "renal_clean.xlsx", index=False)
```
