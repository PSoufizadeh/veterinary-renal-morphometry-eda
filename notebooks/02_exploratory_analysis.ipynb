# 02 | Exploratory data analysis

Sanitised and reorganised from the original EDA notebook. D-Tale is the exploratory tool described in the associated article. Profiling and Sweetviz are additional exploratory utilities present in the source notebook; they are not represented as formal study analyses. See [the article](https://doi.org/10.1111/vru.70287) for findings.

```python
from pathlib import Path
import pandas as pd

# Works from the repository root or notebooks/.
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data"
OUTPUT = ROOT / "outputs"
```

## Load prepared measurements

```python
df = pd.read_excel(DATA / "renal_clean.xlsx")
# Label-encoded nominal variables must not be interpreted as continuous.
nominal = ["Species", "Breed", "Sex", "Reproductive status"]
for column in nominal:
    if column in df:
        df[column] = df[column].astype("category")
```

## D-Tale interactive exploration

Explore distributions, descriptive statistics and Predictive Power Score (PPS) in the local interface. PPS is descriptive and exploratory, not a validated prediction model or inferential test. Do not expose the local service or publish its outputs.

```python
import dtale
session = dtale.show(df, host="127.0.0.1")
session
```

## Optional automated profiling

Modern replacement for the original pandas_profiling import. Generated reports may contain sensitive data and must remain local.

```python
# Optional: enable only in a private environment.
RUN_OPTIONAL_REPORTS = False
if RUN_OPTIONAL_REPORTS:
    from ydata_profiling import ProfileReport
    OUTPUT.mkdir(parents=True, exist_ok=True)
    ProfileReport(df, title="Renal morphometry EDA", explorative=True).to_file(
        OUTPUT / "renal_profile.html"
    )
```

## Optional Sweetviz report

```python
if RUN_OPTIONAL_REPORTS:
    import sweetviz as sv
    OUTPUT.mkdir(parents=True, exist_ok=True)
    report = sv.analyze(df)
    report.show_html(str(OUTPUT / "renal_sweetviz.html"), open_browser=False)
```
